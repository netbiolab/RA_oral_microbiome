## Filter out samples from each count matrix for RA patients analysis

### Change sample name for different sample names matrix

```python
import pandas as pd
# Load Matrix and metadata
bacteria_matrix = pd.read_csv('./99_Original_Data/bacteria_final_normalized_count.tsv', sep = '\t', index_col=0)
fungi_matrix = pd.read_csv('./99_Original_Data/fungi_final_normalized_count.tsv', sep = '\t', index_col=0)
virus_matrix = pd.read_csv('./99_Original_Data/virus_final_normalized_count.tsv', sep = '\t', index_col=0)
MetaCyc_pathway = pd.read_csv('./99_Original_Data/pathabundance_merged_filtered_final.tsv', sep = '\t', index_col=0)
metadata = pd.read_csv('./01_Processed_Data/cleaned_metadata.tsv', sep = '\t', index_col=1)
```

```python
virus_matrix = virus_matrix.T
virus_matrix.fillna(0, inplace=True)
fungi_matrix.index = fungi_matrix.index.str.replace('_bracken', '')
MetaCyc_pathway = MetaCyc_pathway.T
MetaCyc_pathway.index = MetaCyc_pathway.index.str.replace('_Abundance', '')
```

```python
# Make relative abundance matrix for all matrices
bacteria_matrix_rel_abundance = bacteria_matrix.div(bacteria_matrix.sum(axis=1), axis=0)
fungi_matrix_rel_abundance = fungi_matrix.div(fungi_matrix.sum(axis=1), axis=0)
virus_matrix_rel_abundance = virus_matrix.div(virus_matrix.sum(axis=1), axis=0)
MetaCyc_pathway_rel_abundance = MetaCyc_pathway.div(MetaCyc_pathway.sum(axis=1), axis=0)
# Save the matrices
bacteria_matrix_rel_abundance.to_csv('./01_Processed_Data/bacteria_matrix_rel_abundance.tsv', sep='\t', index=True)
fungi_matrix_rel_abundance.to_csv('./01_Processed_Data/fungi_matrix_rel_abundance.tsv', sep='\t', index=True)
virus_matrix_rel_abundance.to_csv('./01_Processed_Data/virus_matrix_rel_abundance.tsv', sep='\t', index=True)
MetaCyc_pathway_rel_abundance.to_csv('./01_Processed_Data/MetaCyc_pathway_rel_abundance.tsv', sep='\t', index=True)
```

```python
# filter our only samples with follow-up DAS28ESR value (DAS_28_ESR.1)
follow_up_samples = metadata[metadata['DAS_28_ESR.1'].notna()].index
# Filter matrices to keep only follow-up samples
bacteria_matrix = bacteria_matrix.loc[follow_up_samples]
fungi_matrix = fungi_matrix.loc[follow_up_samples]
virus_matrix = virus_matrix.loc[follow_up_samples]
MetaCyc_pathway = MetaCyc_pathway.loc[follow_up_samples]
# Do the same also for relative abundance matrices
bacteria_matrix_rel_abundance = bacteria_matrix_rel_abundance.loc[follow_up_samples]
fungi_matrix_rel_abundance = fungi_matrix_rel_abundance.loc[follow_up_samples]
virus_matrix_rel_abundance = virus_matrix_rel_abundance.loc[follow_up_samples]
MetaCyc_pathway_rel_abundance = MetaCyc_pathway_rel_abundance.loc[follow_up_samples]
# Save the filtered matrices
bacteria_matrix.to_csv('./01_Processed_Data/bacteria_matrix_filtered.tsv', sep='\t', index=True)
fungi_matrix.to_csv('./01_Processed_Data/fungi_matrix_filtered.tsv', sep='\t', index=True)
virus_matrix.to_csv('./01_Processed_Data/virus_matrix_filtered.tsv', sep='\t', index=True)
MetaCyc_pathway.to_csv('./01_Processed_Data/MetaCyc_pathway_filtered.tsv', sep='\t', index=True)
bacteria_matrix_rel_abundance.to_csv('./01_Processed_Data/bacteria_matrix_rel_abundance_filtered.tsv', sep='\t', index=True)
fungi_matrix_rel_abundance.to_csv('./01_Processed_Data/fungi_matrix_rel_abundance_filtered.tsv', sep='\t', index=True)
virus_matrix_rel_abundance.to_csv('./01_Processed_Data/virus_matrix_rel_abundance_filtered.tsv', sep='\t', index=True)
MetaCyc_pathway_rel_abundance.to_csv('./01_Processed_Data/MetaCyc_pathway_rel_abundance_filtered.tsv', sep='\t', index=True)
```
